```python
from pathlib import Path

from ultralytics import YOLO

PROJECT_ROOT = Path.cwd()
DATA_YAML = PROJECT_ROOT / "dataset" / "data.yaml"

if not DATA_YAML.is_file():
    raise FileNotFoundError(f"Dataset config not found: {DATA_YAML}. Run this notebook from the project root.")

model = YOLO("yolo11n.pt")
train_results = model.train(
    data=str(DATA_YAML),
    epochs=50,
    imgsz=640,
    batch=16,
    project=str(PROJECT_ROOT / "runs" / "detect"),
    name="ppe",
)
```

```python
from pathlib import Path

from ultralytics import YOLO

PROJECT_ROOT = Path.cwd()
WEIGHTS = PROJECT_ROOT / "runs" / "detect" / "train2" / "weights" / "best.pt"
TEST_IMAGES = PROJECT_ROOT / "dataset" / "test" / "images"

if not WEIGHTS.is_file():
    raise FileNotFoundError(f"Trained weights not found: {WEIGHTS}")

model = YOLO(str(WEIGHTS))
predictions = model.predict(source=str(TEST_IMAGES), save=True, imgsz=640, conf=0.25)
if predictions:
    predictions[0].show()
```

```python
from pathlib import Path

from ultralytics import YOLO

PROJECT_ROOT = Path.cwd()
WEIGHTS = PROJECT_ROOT / "runs" / "detect" / "train2" / "weights" / "best.pt"
VIDEO = PROJECT_ROOT / "A building site labourer - 20 years ago Vs Today.mp4"

if not WEIGHTS.is_file():
    raise FileNotFoundError(f"Trained weights not found: {WEIGHTS}")
if not VIDEO.is_file():
    raise FileNotFoundError(f"Video not found: {VIDEO}")

model = YOLO(str(WEIGHTS))
video_results = model.predict(source=str(VIDEO), save=True, imgsz=640, conf=0.25)
if video_results:
    video_results[0].show()
```
